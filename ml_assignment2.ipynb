```python
import pandas as pd

data = {
    'Player':  ['Rohit', 'Virat', 'Bumrah', 'Jadeja'],
    'Runs':    [9600, 13000, 400, 2800],
    'Active':  [True, True, True, True],
}
df = pd.DataFrame(data)
print(df.shape)
print(df['Runs'][1])
print(df[df['Runs'] > 5000]['Player'].tolist())
print(df.dtypes['Active'])
```

```python
data = {
    'City':    ['Mumbai', 'Delhi', None, 'Pune', 'Delhi'],
    'Temp':    [32.5, 28.0, 31.0, '29.5', 45.0],
    'Season':  ['Summer','Winter','summer','Winter','Summer'],
}
df = pd.DataFrame(data)
df["City"] = df["City"].fillna(df["City"].mode()[0])
print(df)
```

```python
df["Temp"] = pd.to_numeric(df["Temp"])
print(df)
```

```python
df["Season"] = df["Season"].str.strip().str.title()
print(df)
```

```python
products = pd.DataFrame({
    'Product':  ['Pen','Book','Ruler','Bag','Pencil','Eraser'],
    'Price':    [10, 150, 25, 500, 5, 8],
    'Stock':    [200, 50, 80, 30, 300, 150],
    'Category': ['Stationery','Stationery','Stationery','Bag','Stationery','Stationery'],
})
print(products[products["Price"] < 50])
```

```python
products["Value"] = products["Stock"] * products["Price"]
print(products)
```

```python
print(products.groupby("Category")["Value"].sum())
```

```python
products = products.sort_values("Price", ascending=False)
print(products)
```

```python
products["Expensive"] = products["Price"].apply(lambda x: True if x > 100 else False)
print(products)
```

```python
messy = pd.DataFrame({
    'Student': ['Aryan', 'Priya', None, 'Kabir', 'Aryan'],
    'Score':   [88, 95, 72, None, 88],
    'Grade':   ['B', 'A', 'C', 'B', 'B'],
    'City':    ['Pune', 'delhi', 'PUNE', 'Mumbai', 'Pune'],
})
messy["Student"] = messy["Student"].fillna("Unknown")
print(messy)
```

```python
messy = messy.drop_duplicates()
print(messy)
```

```python
messy["Score"] = messy["Score"].fillna(messy["Score"].mean())
print(messy)
```

```python
messy["City"] = messy["City"].str.strip().str.title()
print(messy)
```
